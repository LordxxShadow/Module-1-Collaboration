# Exercise 1: Sort an Array of 0s, 1s, and 2s

This exercise demonstrates an object-oriented programming problem involving sorting an array containing only 0s, 1s, and 2s.

```python
class Solution:
    def sort012(self, arr):
        low = 0
        mid = 0
        high = len(arr) - 1

        while mid <= high:
            if arr[mid] == 0:
                arr[low], arr[mid] = arr[mid], arr[low]
                low += 1
                mid += 1
            elif arr[mid] == 1:
                mid += 1
            else:
                arr[mid], arr[high] = arr[high], arr[mid]
                high -= 1


arr = [0, 1, 2, 0, 1, 2]

solution = Solution()
solution.sort012(arr)

print("Sorted array:", arr)
```

### Explanation

The `Solution` class contains the `sort012()` method. The method sorts an array containing only 0s, 1s, and 2s without using Python's built-in sorting function. It uses three pointers: `low`, `mid`, and `high` to place the values in the correct order.

# Exercise 2: Binary Search

This exercise demonstrates functional programming using a lambda expression to perform a binary search.

```python
# Binary Search

numbers = [2, 4, 6, 8, 10, 12, 14]
target = 10


def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1


search = lambda arr, target: binary_search(arr, target)

result = search(numbers, target)

print("Array:", numbers)
print("Target:", target)
print("Index:", result)
```

### Explanation

The `binary_search()` function searches a sorted array by checking the middle value and eliminating half of the remaining search area. The lambda expression calls the binary search function. If the target is found, the function returns its index. Otherwise, it returns `-1`.
